```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("Automobile_data (2).csv")

# Display first 5 rows
print(df.head())

# Dataset Information
print(df.info())

# Statistical Summary
print(df.describe())

# Set Plot Style
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['horsepower'].dropna(), bins=15, color='skyblue', edgecolor='black')
plt.title("Distribution of Horsepower")
plt.xlabel("Horsepower")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['horsepower'], df['price'], color='red')
plt.title("Horsepower vs Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='body-style', data=df)
plt.title("Count of Body Styles")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='fuel-type', y='price', data=df)
plt.title("Fuel Type vs Price")
plt.show()
```

```python
body = df['body-style'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(body, labels=body.index, autopct='%1.1f%%', startangle=90)
plt.title("Body Style Distribution")
plt.show()
```

```python
df_sorted = df.sort_values(by='horsepower')

plt.figure(figsize=(8,4))
plt.plot(df_sorted['horsepower'], df_sorted['price'], color='green')
plt.title("Horsepower vs Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(10,8))

numeric_df = df.select_dtypes(include=['int64','float64'])

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[['horsepower','price','engine-size','curb-weight']])
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(x='fuel-type', y='horsepower', data=df)
plt.title("Fuel Type vs Horsepower")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['price'], kde=True)
plt.title("Price Distribution")
plt.show()
```
